##Import libraries and configure paths

```python
import pandas as pd
import numpy as np
import os

DATA_PATH = "/content/drive/MyDrive/trendpulse/trendpulse/part2_numpy_pandas/data"

COUNTRIES = ["US", "GB", "IN"]

print("TrendPulse - Part 2")
print("NumPy & Pandas Data Wrangling")
```

##Load the three datasets

```python
us_df = pd.read_csv(os.path.join(DATA_PATH, "Copy of USvideos.csv"))
gb_df = pd.read_csv(os.path.join(DATA_PATH, "Copy of GBvideos.csv"))
in_df = pd.read_csv(os.path.join(DATA_PATH, "Copy of INvideos.csv"))

us_df["country"] = "US"
gb_df["country"] = "GB"
in_df["country"] = "IN"

print("US:", us_df.shape)
print("GB:", gb_df.shape)
print("IN:", in_df.shape)
```

##NumPy arrays

```python
views = us_df["views"].to_numpy()
likes = us_df["likes"].to_numpy()
dislikes = us_df["dislikes"].to_numpy()
comments = us_df["comment_count"].to_numpy()

arrays = {
    "Views": views,
    "Likes": likes,
    "Dislikes": dislikes,
    "Comments": comments
}

for name, arr in arrays.items():
    print(name)
    print("Shape:", arr.shape)
    print("Data type:", arr.dtype)
    print()
```

##Engagement score using NumPy

```python
views = df["views"].to_numpy(dtype=float)
likes = df["likes"].to_numpy(dtype=float)
dislikes = df["dislikes"].to_numpy(dtype=float)
comments = df["comment_count"].to_numpy(dtype=float)

like_ratio = np.divide(
    likes,
    likes + dislikes,
    out=np.zeros_like(likes),
    where=(likes + dislikes) != 0
)

comment_rate = np.divide(
    comments,
    views,
    out=np.zeros_like(comments),
    where=views != 0
)

def normalize(array):
    minimum = np.min(array)
    maximum = np.max(array)

    if maximum == minimum:
        return np.zeros_like(array)

    return (array - minimum) / (maximum - minimum)

df["engagement_score"] = (
    0.35 * normalize(views) +
    0.35 * normalize(likes) +
    0.15 * normalize(comments) +
    0.10 * like_ratio +
    0.05 * normalize(comment_rate)
)

df["log_views"] = np.log1p(views)

threshold = df["engagement_score"].quantile(0.90)

df["top_10_percent"] = df["engagement_score"] >= threshold

print("Missing engagement scores:", df["engagement_score"].isna().sum())
print("Top 10% records:", df["top_10_percent"].sum())
print("Engagement threshold:", threshold)
```

##Create combined DataFrame and calculate scores

```python
df = pd.concat(
    [us_df, gb_df, in_df],
    ignore_index=True
)

df["engagement_score"] = np.nan

df.loc[us_df.index, "engagement_score"] = engagement_score

print("Combined shape:", df.shape)
print(df.head())
```

```python
views_all = df["views"].to_numpy(dtype=float)
likes_all = df["likes"].to_numpy(dtype=float)
dislikes_all = df["dislikes"].to_numpy(dtype=float)
comments_all = df["comment_count"].to_numpy(dtype=float)

like_ratio_all = np.divide(
    likes_all,
    likes_all + dislikes_all,
    out=np.zeros_like(likes_all),
    where=(likes_all + dislikes_all) != 0
)

comment_rate_all = np.divide(
    comments_all,
    views_all,
    out=np.zeros_like(comments_all),
    where=views_all != 0
)

def normalize(array):
    minimum = np.min(array)
    maximum = np.max(array)

    if maximum == minimum:
        return np.zeros_like(array, dtype=float)

    return (array - minimum) / (maximum - minimum)

views_norm_all = normalize(views_all)
likes_norm_all = normalize(likes_all)
comments_norm_all = normalize(comments_all)
comment_rate_norm_all = normalize(comment_rate_all)

df["engagement_score"] = (
    0.35 * views_norm_all +
    0.35 * likes_norm_all +
    0.15 * comments_norm_all +
    0.10 * like_ratio_all +
    0.05 * comment_rate_norm_all
)

df["log_views"] = np.log1p(views_all)

threshold = df["engagement_score"].quantile(0.90)

df["top_10_percent"] = df["engagement_score"] >= threshold

print("Top 10% records:", df["top_10_percent"].sum())
```

##Reshape and broadcasting

```python
daily_counts = (
    df.groupby(["trending_date", "country"])
      .size()
      .unstack(fill_value=0)
)

daily_matrix = daily_counts.to_numpy()

print("Original shape:", daily_matrix.shape)

flattened = daily_matrix.flatten()

print("Flattened shape:", flattened.shape)

reshaped = flattened.reshape(daily_matrix.shape)

print("Reshaped shape:", reshaped.shape)

# Country share of each day's total
daily_totals = daily_matrix.sum(axis=1, keepdims=True)

country_share = np.divide(
    daily_matrix,
    daily_totals,
    out=np.zeros_like(daily_matrix, dtype=float),
    where=daily_totals != 0
)

# Daily value relative to each country's average
country_average = daily_matrix.mean(axis=0, keepdims=True)

relative_to_average = np.divide(
    daily_matrix,
    country_average,
    out=np.zeros_like(daily_matrix, dtype=float),
    where=country_average != 0
)

print("Country share:")
print(country_share[:5])

print("Relative to country average:")
print(relative_to_average[:5])
```

##Inspect and clean the combined dataset

```python
print("Dataset shape:", df.shape)

print("\nFirst five rows:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nStatistical description:")
display(df.describe())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isna().sum())
```

##Convert data types and dates

```python
numeric_columns = [
    "views",
    "likes",
    "dislikes",
    "comment_count",
    "category_id"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

df["trending_date"] = pd.to_datetime(
    df["trending_date"],
    format="%y.%d.%m",
    errors="coerce"
)

df["publish_time"] = pd.to_datetime(
    df["publish_time"],
    errors="coerce",
    utc=True
)

print("Updated data types:")
print(df.dtypes)

print("\nDate parsing failures:")
print("Trending date:", df["trending_date"].isna().sum())
print("Publish time:", df["publish_time"].isna().sum())
```

##Missing values and duplicates

```python
missing_count = df.isna().sum()

missing_percentage = (
    df.isna().mean() * 100
)

missing_report = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

display(missing_report[missing_report["Missing Count"] > 0])

# Fill missing descriptions
df["description"] = df["description"].fillna("")

# Fill missing numeric values
numeric_columns = [
    "views",
    "likes",
    "dislikes",
    "comment_count"
]

df[numeric_columns] = df[numeric_columns].fillna(0)

# Remove rows with missing essential identifiers or dates
df = df.dropna(
    subset=["video_id", "trending_date", "country"]
)

# Remove duplicate records
before = len(df)

df = df.drop_duplicates(
    subset=["video_id", "trending_date", "country"]
)

after = len(df)

print("Duplicate rows removed:", before - after)
print("Final shape:", df.shape)
```

##Load categories and merge

```python
categories_df = pd.read_csv(
    os.path.join(DATA_PATH, "Copy of categories.csv")
)

print(categories_df.head())
print(categories_df.columns)
```

```python
before_merge = len(df)

df = df.merge(
    categories_df,
    on="category_id",
    how="left"
)

print("Records before merge:", before_merge)
print("Records after merge:", len(df))

print(df[["video_id", "category_id", "category_name"]].head())
```

##Filtering

```python
# A: Videos with 1 million+ views and more than 50,000 likes
popular_videos = df[
    (df["views"] >= 1_000_000) &
    (df["likes"] > 50_000)
]

display(
    popular_videos[
        ["title", "country", "views", "likes"]
    ].head(20)
)
```

```python
# B: Top 20 records by engagement score
top_20 = df.sort_values(
    by="engagement_score",
    ascending=False
).head(20)

display(
    top_20[
        ["title", "country", "views", "likes", "engagement_score"]
    ]
)
```

```python
# C: Music or Gaming above overall median views
median_views = df["views"].median()

music_gaming = df[
    (df["category_name"].isin(["Music", "Gaming"])) &
    (df["views"] > median_views)
]

display(
    music_gaming[
        ["title", "category_name", "country", "views"]
    ].head(20)
)

# loc example
display(df.loc[0:4, ["title", "views", "likes"]])

# iloc example
display(df.iloc[0:5, 0:5])
```

##Grouping and aggregation

```python
category_summary = df.groupby("category_name").agg(
    record_count=("video_id", "count"),
    average_views=("views", "mean"),
    average_likes=("likes", "mean"),
    average_engagement=("engagement_score", "mean")
).sort_values(
    by="average_views",
    ascending=False
)

display(category_summary)
```

```python
country_summary = df.groupby("country").agg(
    record_count=("video_id", "count"),
    average_views=("views", "mean"),
    average_likes=("likes", "mean"),
    average_engagement=("engagement_score", "mean")
)

display(country_summary)
```

```python
# Category counts and percentages
category_counts = df["category_name"].value_counts()

category_percentages = (
    df["category_name"].value_counts(normalize=True) * 100
)

print("Category counts:")
print(category_counts)

print("\nCategory percentages:")
print(category_percentages)
```

```python
# Best publishing weekday
df["publish_weekday"] = df["publish_time"].dt.day_name()

weekday_views = df.groupby(
    "publish_weekday"
)["views"].mean().sort_values(ascending=False)

print(weekday_views)

# Best publishing hour
df["publish_hour"] = df["publish_time"].dt.hour

hour_views = df.groupby(
    "publish_hour"
)["views"].mean().sort_values(ascending=False)

print(hour_views)
```

##Time to trend

```python
lifespan = df.groupby(
    ["country", "video_id", "title", "category_name"]
)["trending_date"].nunique().reset_index()

lifespan.rename(
    columns={"trending_date": "trending_lifespan"},
    inplace=True
)

print("10 longest-lasting videos:")

display(
    lifespan.sort_values(
        by="trending_lifespan",
        ascending=False
    ).head(10)
)
```

```python
df["days_to_trend"] = (
    df["trending_date"] -
    df["publish_time"].dt.tz_localize(None)
).dt.total_seconds() / 86400

print(df["days_to_trend"].describe())
```

##Trending lifespan

```python
lifespan = df.groupby(
    ["country", "video_id", "title", "category_name"]
)["trending_date"].nunique().reset_index()

lifespan.rename(
    columns={"trending_date": "trending_lifespan"},
    inplace=True
)

print("10 longest-lasting videos:")

display(
    lifespan.sort_values(
        by="trending_lifespan",
        ascending=False
    ).head(10)
)
```

```python
# Average lifespan by category
category_lifespan = lifespan.groupby(
    "category_name"
)["trending_lifespan"].mean().sort_values(ascending=False)

display(category_lifespan)
```

##Combine per-video and per-channel summaries


```python
video_summary = df.groupby(
    ["country", "video_id"]
).agg(
    total_views=("views", "max"),
    average_engagement=("engagement_score", "mean")
).reset_index()

channel_summary = df.groupby(
    ["country", "channel_title"]
).agg(
    trending_videos=("video_id", "nunique"),
    total_views=("views", "sum")
).reset_index()

channel_report = channel_summary.merge(
    df[["country", "channel_title", "video_id"]].drop_duplicates()
      .groupby(["country", "channel_title"])
      .size()
      .reset_index(name="unique_video_count"),
    on=["country", "channel_title"],
    how="left"
)

display(
    channel_report.sort_values(
        by="total_views",
        ascending=False
    ).head(20)
)
```
